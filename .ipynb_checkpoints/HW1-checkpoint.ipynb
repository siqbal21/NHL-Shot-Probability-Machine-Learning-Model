## Part 1.1: Manual Data Entry (25 points)


How would you manually track temperature changes throughout the day without an automated weather station? Go to this website: [this website](https://www.weather.gov/) and look up your local area's current temperature. You are going to write a function that takes manual input of temperature readings (entered by you), stores all readings in a list, stops once you enter 15 values or four consecutive increasing values (whichever comes first), and returns the temperature data.
For this problem:

- Call your function `temp_tracker`
- Make sure that you use the `input()` option within your function to accept entries
- Make sure you are saving all entries in a list
- Use `if-else` and a `while` loop
- When you have written the function, call:

`temperature_log = temp_tracker()`

`temperature_log`

and manually check the weather website to enter the current temperature (refresh the page every few minutes for updates).

**Note**: There may be multiple ways to get this function to work, but to receive full points, your function **must** have the five components listed above (there may be more efficient ways to do it, but I want you to show me you know how to use *these methods*).

```python
# 1.
def temp_tracker():
    '''
    takes manual input of temperature readings and stores
    the readings into a list. Terminates when 15 temperature
    readings are recorded or if the last 4 readings are all increasing
    Args:
    
    Returns:
        list of floats representing the inputted temperature readings
    '''
    readings = []
    last_four_increasing = False
    while len(readings) < 15 and last_four_increasing == False:
        readings.append(float(input("Input the next temperature reading")))
    
        if len(readings) >= 4:
            last_four = readings[-4:]
            if last_four[0] < last_four[1] < last_four[2] < last_four[3]:
                last_four_increasing = True
            else:
                last_four_increasing = False
    
    return readings


temperature_log = temp_tracker()
temperature_log
```

# Adding Arrays Explanation

Adding array_a and t_array_b together results in an error because the two arrays do not share the same dimensions. array_a is a 3 by 4 array while t_array_b is a 3 by 5 array. In order for two arrays to be added together, they must share the same shape, and because t_array_b has one more column than array_a the two are not compatible for addition.

```python
t_array_b = t_array_b[:, :-1]

array_a + t_array_b
```

# Adding Arrays Explanation Cont.
Now that both array_a and t_array_b are the same size, the two arrays can be added together element by element. After removing the last column of t_array_b, both array_a and t_array_b share the same shape as 3 by 4 arrays, therefore they can be added together.

```python
# 1B.
index_highest = pokedex["BMI"].idxmax()
index_lowest = pokedex["BMI"].idxmin()
print(f"Max BMI in pokedex dataframe: {index_highest}")
print(f"Min BMI in pokedex dataframe: {index_lowest}")

highest_BMI = pokedex.loc[index_highest, "Pokemon"]
lowest_BMI = pokedex.loc[index_lowest, "Pokemon"]
print(f"Pokemon with Highest BMI: {highest_BMI}")
print(f"Pokemon with Lowest BMI: {lowest_BMI}")
```

```python
# 2.
pokedex_defense62_nonwater = pokedex[(pokedex['Defense'] >= 62) & (pokedex['MainType'] != 'Water')]

print(pokedex_defense62_nonwater.head)
```

# Part 4.2: Manipulating dataset (25 points)

1. Extract the `total_bill` column as a NumPy array. Slice the array to get elements from index 10 to 20 (inclusive). Print the sliced array.

2. Using `iloc`, select rows 5-15 (inclusive) and columns 1-3 (inclusive), and display the result. Using `loc`, select all rows where `time` is `Dinner` and display only the `total_bill` and `tip` columns.

3. Create a new column called `tip_percentage` that calculates the tip as a percentage of the total bill. Display the new columns after calculation

4. Create a new DataFrame with a single column called `bill_category` that categorizes each bill as `Low` (< \$15), `Medium` (\$15-
\$25), or `High` (> \$25). Use pd.concat to add this new column to the original dataset. Save the modified dataset to a CSV file named `modified_tips.csv`. Print the head of the `modified_tips.csv`


```python
# 1.
# Convert series into numpy array
total_bill_arr = tips["total_bill"].to_numpy()
total_bill_arr_slice = total_bill_arr[10:21]

print(total_bill_arr_slice)
```

```python
# 2.
rows_and_cols1 = tips.iloc[5:16, 1:4]

rows_and_cols2 = tips.loc[tips["time"] == "Dinner", ["total_bill", "tip"]]

print(rows_and_cols1)
print("\n")
print(rows_and_cols2)
```

```python
# 3.
tip_percent = round(100 * tips["tip"] / tips["total_bill"], 2)

tips["tip_percentage"] = tip_percent
print("tip_percentage column:")
print(tips["tip_percentage"].head)
```

```python
# 4.
bill_category = pd.DataFrame(pd.cut(tips["total_bill"], 
                                    bins = [0, 15, 25, float("inf")],
                                    labels = ["Low", "Medium", "High"]))

bill_category.columns = ["bill_category"]

# Concatenate the two different data frames together
tips = pd.concat([tips, bill_category], axis = 1)

# Save the combined data frames into csv file
tips.to_csv("modified_tips.csv", index = False)
modified_tips = pd.read_csv("modified_tips.csv")
modified_tips.head
```
